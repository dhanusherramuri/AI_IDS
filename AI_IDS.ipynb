```python
# import pandas as pd
# import os

# for f in files[:5]:
#     df = pd.read_csv(f)
#     print(f.split("/")[-1])
#     print(df["Label"].unique())
#     print("-" * 50)

import pandas as pd
import glob
import os

# Path to your dataset folder
dataset_path = "/content/drive/MyDrive/MERGED_CSV/*.csv"

# Get all CSV files
files = sorted(glob.glob(dataset_path))

print(f"Total files found: {len(files)}")

# Output file
output_file = "/content/drive/MyDrive/AI_IDS/DATASETS/MERGED_CSV_Sampled_5Percent.csv"

first_file = True

for file in files:

    print(f"\nProcessing: {os.path.basename(file)}")

    # Read one CSV
    df = pd.read_csv(file)

    # Take 5% sample from every attack class (stratified sampling)
    sampled_df = (
        df.groupby("Label", group_keys=False)
          .apply(lambda x: x.sample(frac=0.05, random_state=42))
          .reset_index(drop=True)
    )

    print(f"Original Shape : {df.shape}")
    print(f"Sampled Shape  : {sampled_df.shape}")

    # Write first file with header
    if first_file:
        sampled_df.to_csv(output_file, index=False)
        first_file = False
    else:
        # Append remaining files without header
        sampled_df.to_csv(
            output_file,
            mode='a',
            header=False,
            index=False
        )

    # Free RAM
    del df
    del sampled_df

print("\n✅ Sampling completed successfully!")
print(f"Sampled dataset saved at:\n{output_file}")
```

```python
#*******FEATURE SELECTION USING PEARSON CORRELATION COEFFICIENT**************

import pandas as pd
import numpy as np

df = pd.read_csv("/content/drive/MyDrive/AI_IDS/DATASETS/MERGED_CSV_Clean.csv")
print(df.shape)

X= df.drop("Label" , axis = 1)
y = df["Label"]

print(X.shape)

corr_matrix = X.corr(method = 'pearson')
print(corr_matrix.shape)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(18,15))

sns.heatmap(
    corr_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix")

plt.show()
```

```python
#DECIDING ON WHAT ARE THE COLUMNS THAT ARE TO BE REMOVED USING PCC

threshold = 0.90

upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

to_drop = [
    column
    for column in upper.columns
    if any(abs(upper[column]) > threshold)
]

print("Features to Remove:")
print(to_drop)

print("\nTotal Features to Remove:", len(to_drop))
```

```python
#DROPPING THE REDUNDANT FEATURES

df_selected = df.drop(columns=to_drop)

print("Original Shape :", df.shape)
print("Selected Shape :", df_selected.shape)
```

```python
#VERIFYING THE NEW CORRELATION MATRIX

new_corr = df_selected.drop("Label", axis=1).corr()

plt.figure(figsize=(15,12))

sns.heatmap(
    new_corr,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix After PCC")

plt.show()
```

```python
#LABEL ENCODING THE DATA

import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/AI_IDS/DATASETS/MERGED_CSV_PCC_MinMax.csv")

print(df.shape)

#CREATING A BINARY DATASET
binary_df = df.copy()

binary_df["Label"] = binary_df["Label"].apply(
    lambda x: 0 if x == "BENIGN" else 1
)

print(binary_df["Label"].value_counts())

```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

multi_df = df.copy()

multi_df["Label"] = encoder.fit_transform(multi_df["Label"])

mapping = pd.DataFrame({
    "Attack Type": encoder.classes_,
    "Encoded Label": range(len(encoder.classes_))
})

mapping


#MAPPING WILL BE HELPFUL WHEN CONNECTING IT TO THE DASHBOARD
mapping.to_csv(
    "/content/drive/MyDrive/AI_IDS/DATASETS/Label_Mapping.csv",
    index=False
)
```
